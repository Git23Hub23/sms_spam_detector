```python
import pandas as pd
import re

df = pd.read_csv("../data/spam.csv", encoding="latin-1")
df = df[["v1", "v2"]]
df.columns = ["label", "message"]
df = df.drop_duplicates().reset_index(drop=True)
print(df.shape)
```

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

def clean_text(text):
    text = text.lower()                      # make everything small letters
    text = re.sub(r"[^a-z\s]", " ", text)    # remove punctuation and numbers
    words = text.split()                     # split into words
    words = [stemmer.stem(w) for w in words
             if w not in ENGLISH_STOP_WORDS and len(w) > 2]
    return " ".join(words)
```

```python
sample = df["message"][2]
print("BEFORE:", sample)
print("AFTER: ", clean_text(sample))
```

```python
sample = df["message"][23]
print("BEFORE:", sample)
print("AFTER: ", clean_text(sample))
```

```python
df["clean"] = df["message"].apply(clean_text)
df[["message", "clean"]].head(10)
```

```python
print((df["clean"].str.strip() == "").sum())
```

```python
df = df[df["clean"].str.strip() != ""].reset_index(drop=True)
print(df.shape)
```

```python
df["label_num"] = df["label"].map({"ham": 0, "spam": 1})
df[["label", "label_num", "clean"]].head()
```

```python
df.to_csv("../data/clean.csv", index=False)
print("Saved!")
```
